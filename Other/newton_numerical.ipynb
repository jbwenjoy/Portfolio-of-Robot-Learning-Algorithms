```python
"""
Given a function ln(x) + x^3 = 0, find the root of within (0, 1) with accuracy 1e-6 using the Newton-Raphson method.

"""
import math

def f(x):
    return math.log(x) + x**3

def f_prime(x):
    return 1/x + 3*x**2

def newton_raphson(x0, tol):
    x = x0
    while True:
        x_next = x - f(x) / f_prime(x)
        if abs(x_next - x) < tol:
            return x_next
        x = x_next

if __name__ == "__main__":
    x0 = 0.5
    tol = 1e-6
    root = newton_raphson(x0, tol)
    print(root)
    print(f(root))
```

```python
"""
Given any function (user input), find the root of within (0, 1) with accuracy 1e-6 using the Newton-Raphson method.
If no root exists, return None.

"""

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from sympy.parsing.sympy_parser import parse_expr
from sympy import lambdify

def newton_raphson(f, f_prime, x0, tol):
    x = x0
    while True:
        x_next = x - f(x) / f_prime(x)
        if abs(x_next - x) < tol:
            return x_next
        x = x_next

if __name__ == "__main__":
    x = sp.symbols('x')
    f_expr = input("Enter the function (e.g. x**2 - 4): ")
    print(f"f(x) = {f_expr}")
    f = lambdify(x, parse_expr(f_expr))
    f_prime = lambdify(x, sp.diff(parse_expr(f_expr), x))
    x0 = 0.5
    tol = 1e-6
    root = newton_raphson(f, f_prime, x0, tol)
    if root is not None:
        print(f"x = {root}")
        print(f"f(x) = {f(root)}")
    else:
        print("No root exists in the interval (0, 1).")
```

```python
"""
Given a function f(x) = x^3 - 2x - 5, find the root of within (0, 1) with accuracy 1e-6 using binary search.

"""
import math

def f(x):
    return math.log(x) + x**3

def binary_search(lb, ub, tol):
    while ub - lb > tol:
        c = (lb + ub) / 2
        if f(lb) * f(c) < 0:
            ub = c
        else:
            lb = c
    return (lb + ub) / 2

if __name__ == "__main__":
    lb = 1e-7 # Greater than 0 since log(0) is undefined
    ub = 1
    tol = 1e-6
    root = binary_search(lb, ub, tol)
    print(root)
    print(f(root))
```
